# Environment Configuration Check

This notebook verifies the deep learning environment setup across different devices (Windows, macOS, Linux) and displays system configuration including CUDA, GPU, and installed package versions.

**Use this to validate:**
- OS compatibility (Windows/macOS/Linux)
- Python version
- CUDA & GPU availability
- Key package versions
- System memory and device information


```python
## 1. Import Required Libraries
import sys
import platform
import subprocess
from pathlib import Path
import torch
import numpy as np

print("✓ All imports successful")
```

```python
## 2. Check Python Version
print("=" * 60)
print("PYTHON VERSION")
print("=" * 60)
print(f"Python Version: {sys.version}")
print(f"Python Implementation: {platform.python_implementation()}")
print(f"Python Executable: {sys.executable}")
```

```python
## 3. Detect Operating System
print("\n" + "=" * 60)
print("OPERATING SYSTEM")
print("=" * 60)
os_name = platform.system()
os_release = platform.release()
os_version = platform.version()
print(f"OS: {os_name}")
print(f"Release: {os_release}")
print(f"Version: {os_version}")
print(f"Architecture: {platform.machine()}")

# OS-specific info
if os_name == "Windows":
    print("✓ Windows detected - ensure CUDA Toolkit is installed from NVIDIA")
elif os_name == "Darwin":
    print("✓ macOS detected - check for Metal Performance Shaders (MPS) support")
elif os_name == "Linux":
    print("✓ Linux detected - ensure cuDNN and CUDA drivers are installed")
```

```python
## 4. Check CUDA Availability
print("\n" + "=" * 60)
print("CUDA & GPU")
print("=" * 60)
cuda_available = torch.cuda.is_available()
print(f"CUDA Available: {cuda_available}")

if cuda_available:
    print(f"CUDA Version: {torch.version.cuda}")
    print(f"cuDNN Version: {torch.backends.cudnn.version()}")
    print(f"cuDNN Enabled: {torch.backends.cudnn.enabled}")
else:
    print("⚠ CUDA not available - will use CPU for computations")
    if os_name == "Darwin":
        print("  → On macOS, PyTorch may use Metal Performance Shaders instead")
```

```python
## 5. GPU Information
print("\n" + "=" * 60)
print("GPU DETAILS")
print("=" * 60)

if torch.cuda.is_available():
    num_gpus = torch.cuda.device_count()
    print(f"Number of GPUs: {num_gpus}")
    
    for i in range(num_gpus):
        print(f"\nGPU {i}:")
        print(f"  Name: {torch.cuda.get_device_name(i)}")
        print(f"  Capability: {torch.cuda.get_device_capability(i)}")
        
        # Memory info
        props = torch.cuda.get_device_properties(i)
        total_memory = props.total_memory / (1024 ** 3)  # Convert to GB
        print(f"  Total Memory: {total_memory:.2f} GB")
        
        # Current device
        if i == torch.cuda.current_device():
            print(f"  → Current device")
else:
    print("No GPU detected - compute will run on CPU")
    print(f"PyTorch Device: {torch.device('cpu')}")
```

```python
## 6. Check Installed Package Versions
print("\n" + "=" * 60)
print("KEY PACKAGE VERSIONS")
print("=" * 60)

packages = {
    "NumPy": np,
    "PyTorch": torch,
    "TorchVision": None,  # Handle separately
}

for name, module in packages.items():
    if module is not None:
        version = getattr(module, '__version__', 'Unknown')
        print(f"{name}: {version}")

# TorchVision
try:
    import torchvision
    print(f"TorchVision: {torchvision.__version__}")
except ImportError:
    print("TorchVision: Not installed")

# Additional packages
try:
    import sklearn
    print(f"scikit-learn: {sklearn.__version__}")
except ImportError:
    print("scikit-learn: Not installed")

try:
    import matplotlib
    print(f"Matplotlib: {matplotlib.__version__}")
except ImportError:
    print("Matplotlib: Not installed")

try:
    import jupyter
    print(f"Jupyter: {jupyter.__version__}")
except ImportError:
    print("Jupyter: Not installed")
```

```python
## 7. System Information Summary
import json
from datetime import datetime

print("\n" + "=" * 60)
print("ENVIRONMENT SUMMARY")
print("=" * 60)

summary = {
    "timestamp": datetime.now().isoformat(),
    "operating_system": platform.system(),
    "python_version": platform.python_version(),
    "torch_version": torch.__version__,
    "cuda_available": torch.cuda.is_available(),
    "cuda_version": torch.version.cuda if torch.cuda.is_available() else None,
    "gpu_count": torch.cuda.device_count() if torch.cuda.is_available() else 0,
    "gpu_devices": [torch.cuda.get_device_name(i) for i in range(torch.cuda.device_count())] if torch.cuda.is_available() else [],
}

# Pretty print
for key, value in summary.items():
    print(f"{key}: {value}")

print("\n" + "=" * 60)
if torch.cuda.is_available():
    print("✓ Environment ready for GPU-accelerated deep learning")
else:
    print("⚠ GPU not available - environment will run on CPU")
    
print(f"✓ Project initialized: {os_name} ({platform.python_version()})")
print("=" * 60)
```
